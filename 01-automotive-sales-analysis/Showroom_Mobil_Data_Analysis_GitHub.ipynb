## 4. Data cleaning and preparation

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

url_data = "https://docs.google.com/spreadsheets/d/1GuFkH44urw4ACqt5gsWhi1Scei0pGS2L/export?format=csv"
df = pd.read_csv(url_data)

print("Data berhasil di-load!")
print("Jumlah baris dan kolom:", df.shape)
df.head()

print("Missing value sebelum dibersihkan:")
print(df.isnull().sum())

df["trade_in"] = df["trade_in"].fillna(0)

print("\nJumlah duplikat penuh:", df.duplicated().sum())
print("Jumlah duplikat order_id:", df["order_id"].duplicated().sum())

df = df.drop_duplicates()
df = df.drop_duplicates(subset="order_id")

df["payment_type"] = df["payment_type"].str.strip().str.title()

df["sales_date"] = pd.to_datetime(df["sales_date"])

print("\nInformasi data setelah dibersihkan:")
df.info()

```

## 5. Highest-revenue branch and vehicle-category combinations

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

url_data = "https://docs.google.com/spreadsheets/d/1GuFkH44urw4ACqt5gsWhi1Scei0pGS2L/export?format=csv"
df = pd.read_csv(url_data)

print("Data berhasil di-load!")
print("Jumlah baris dan kolom:", df.shape)
df.head()

df_completed = df[df["status"] == "completed"]

sales_by_combination = (
    df_completed
    .groupby(["branch", "category"])["total_sales"]
    .sum()
)

top_5 = sales_by_combination.sort_values(ascending=False).head(5)

print("5 kombinasi cabang + kategori dengan pendapatan tertinggi:")
print(top_5)

top_5.sort_values().plot(kind="barh")

plt.title("5 Kombinasi Cabang & Kategori dengan Pendapatan Tertinggi")
plt.xlabel("Total Sales")
plt.ylabel("Cabang & Kategori")
plt.show()

```
